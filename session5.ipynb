```python
"""
p>Create 'coaching_marks.ipynb' for a fictional coaching class you name. Make students = ['Meet', 'Dhara', 'Krish', 'Naina', 'Parth', 'Ishita'] and marks = [34, 41, 28, 45, 39, 47] (out of 50). Use append() for 'Zoya' with 36, extend() for ['Rahil', 'Tisha', 'Yash'] with [30, 44, 38], and pop() to remove the student at index 2 and the matching mark. Print both lists and len() after each change; both must end with 9 items. Constraint: Do not use loops.</p>

"""
students = ['Meet', 'Dhara', 'Krish', 'Naina', 'Parth', 'Ishita']
marks=[34,41,28,45,39,47]

print("initial list:")
print(students)
print(marks)
print(len(students))

students.append('zoya')
marks.append(36)

print("\nAfter append:")
print(students)
print(marks)
print(len(students))

students.extend(["Rahil","Tisha","Yash"])
marks.extend([30,44,38])

print("\nAfter extended:")
print(students)
print(marks)
print(len(students))

students.pop(2)
marks.pop(2)
print("\nAfter pop:")
print(students)
print(marks)
print(len(students))
```

```python
"""
<p>Analyse the lists in 'coaching_marks.ipynb' without loops: print the top 3 marks with sorted(marks, reverse=True)[:3], the lowest 3 using slicing, the average rounded to 2 decimals, and the topper's name using marks.index(max(marks)). Then insert 'Dev' with 41 at position 1 in both lists with insert() and print the average again. Expected averages: 39.33, then 39.5.</p>
"""
print("top 3 marks:", sorted(marks, reverse=True)[:3])
print("Lowest 3 Marks:", sorted(marks)[:3])

average = round(sum(marks) / len(marks), 2)
print("Average:", average)

topper_index = marks.index(max(marks))
print("Topper:", students[topper_index])

students.insert(1, 'Dev')
marks.insert(1, 41)

print("\nAfter Inserting Dev:")
print("Students:", students)
print("Marks:", marks)

average = round(sum(marks) / len(marks), 2)
print("New Average:", average)

```
