```python
"""
<p>Create 'bike_types.ipynb' and store a used two-wheeler dealer's quotation as strings: price_text = '48500', accessories_text = '1250.50', insurance_text = '3299.50', rto_text = '1800', km_text = '32000' and owner_count_text = '2'. Convert the four money values to float and km and owners to int, print each value with its type() before and after conversion, and print the on-road total rounded to 2 decimals. It must show 54850.0.</p>

"""
price_text='48500'
accessories_text='1250.50'
insurance_text='3299.50'
rto_text='1800'
km_text='32000'
owner_count_text='2'

print(type(price_text))
print(type(accessories_text))
print(type(insurance_text))
print(type(rto_text))
print(type(km_text))
print(type(owner_count_text))

price=float(price_text)
accessories=float(accessories_text)
insurance=float(insurance_text)
rto=float(rto_text)

km=int(km_text)
owner_count_text=int(owner_count_text)

print(type(price))
print(type(accessories))
print(type(insurance))
print(type(rto))
print(type(km))
print(type(owner_count_text))

total=price+accessories+insurance+rto

print("on-road total:", round(total, 2))
```
