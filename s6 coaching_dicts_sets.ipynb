```python
"""
<p>Finish by building class_profile, a nested dictionary keyed by roll number 101 to 104 whose values hold name, subjects (a set) and fees_paid: 101 Meet {'Maths', 'Physics'} 12000, 102 Dhara {'Maths', 'Biology'} 24000, 103 Naina {'Physics'} 18000, 104 Ishita {'Biology', 'English'} 6000. Add 'Chemistry' to 101's subjects, set 104's fees_paid to 12000, delete roll 103, then print a summary dictionary with total fees (48000) and the set of all subjects (5). Run Restart & Run All and submit.</p>
"""
class_profile = {
    101: {
        "name": "Meet",
        "subjects": {"Maths", "Physics"},
        "fees_paid": 12000,
    },
    102: {
        "name": "Dhara",
        "subjects": {"Maths", "Biology"},
        "fees_paid": 24000,
    },
    103: {
        "name": "Naina",
        "subjects": {"Physics"},
        "fees_paid": 18000,
    },
    104: {
        "name": "Ishita",
        "subjects": {"Biology", "English"},
        "fees_paid": 6000,
    },
}

class_profile[101]["subjects"].add("Chemistry")
class_profile[104]["fees_paid"] = 12000
del class_profile[103]

summary = {
    "total_fees": sum(profile["fees_paid"] for profile in class_profile.values()),
    "subjects": set().union(*(profile["subjects"] for profile in class_profile.values())),
}

print(summary)
```
